```python
import pandas as pd
import numpy as np

# 1. Load Dataset (Hum ek deliberately messy/dirty dataset create kar rahe hain data cleaning practice ke liye)
np.random.seed(42)
data = {
    'CustomerID': [101, 102, 103, 104, 105, 101, 106, 107, 108, 109],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Alice', 'Frank', 'Grace', 'Hannah', 'Ian'],
    'Age': [25, 30, np.nan, 22, 45, 25, 28, np.nan, 35, 40],
    'City': ['New York', 'london', 'Paris', 'New York', 'PARIS', 'New York', 'London', 'Berlin', np.nan, 'Tokyo'],
    'Join_Date': ['2023-01-10', '15-01-2023', '2023-01-12', '2023-01-14', 'invalid_date', '2023-01-10', '2023-01-18', '2023-01-19', '2023-01-20', '2023-01-21'],
    'Purchase_Amount': [250.5, 300.0, 150.75, np.nan, 500.0, 250.5, 120.0, 430.0, 220.0, np.nan]
}

df = pd.DataFrame(data)

print("--- Original Messy Dataset ---")
display(df)

# 2. Data Quality Report (Null counts, duplicates, data types)
print("\n--- Data Quality Report ---")
print("Null Values per Column:")
print(df.isnull().sum())
print("\nDuplicate Rows Count:", df.duplicated().sum())

# 3. Duplicate Removal
df = df.drop_duplicates()
print(f"\n--- After Removing Duplicates: {len(df)} rows remaining ---")

# 4. Missing Data Handling
# Impute Age with median, Purchase_Amount with mean, and City with 'Unknown'
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Purchase_Amount'] = df['Purchase_Amount'].fillna(df['Purchase_Amount'].mean())
df['City'] = df['City'].fillna('Unknown')

# 5. Standardisation (Fixing inconsistent text formatting in City and Date)
df['City'] = df['City'].str.title() # e.g., 'london' or 'PARIS' -> 'London', 'Paris'
df['Join_Date'] = pd.to_datetime(df['Join_Date'], errors='coerce') # Invalid dates turn to NaT
df['Join_Date'] = df['Join_Date'].fillna(df['Join_Date'].mode()[0]) # Fill invalid dates with mode

print("\n--- Cleaned Dataset ---")
display(df)
print("\nFinal Null Check:")
print(df.isnull().sum())
```
